```python
import pandas as pd

inventory = pd.read_csv('../data/processed/inventory_master.csv')

products = pd.read_csv('../data/processed/products.csv')

branches = pd.read_csv('../data/processed/branches.csv')
```

```python
inventory_curated = (
    inventory
    .merge(products, on='product_id', how='left')
    .merge(branches, on='branch_id', how='left')
)

inventory_curated.head()
```

```python
print("Shape:")
print(inventory_curated.shape)

print("\nMissing Values:")
print(inventory_curated.isnull().sum())
```

```python
inventory_curated.to_csv('../data/curated/inventory_curated.csv',
    index=False
)

print("inventory_curated.csv saved!")
```
