```python
import pandas as pd
so_header= pd.read_csv('../data/raw/sales_orders_header.csv')
print(so_header.head())
```

```python
print("===INFO===")
print(so_header.info())

print("===MISSING VALUES===")
print(so_header.isnull().sum())

print("\n=== SHAPE ===")
print(so_header.shape)

print("===SAMPLE DATA===")
print(so_header.sample(5))

```

```python
text_cols = [
    'so_id',
    'customer_id',
    'branch_id',
    'order_status',
    'payment_terms',
    'sales_channel'
]

for col in text_cols:
    so_header[col] = (
        so_header[col]
        .str.upper()
        .str.strip()
    )

so_header.head()
```

```python
so_header['order_date'] = pd.to_datetime(so_header['order_date'],errors='coerce')

so_header['delivery_date'] = pd.to_datetime(so_header['delivery_date'],errors='coerce')

so_header[['order_date', 'delivery_date']].head()
```

```python
so_header.drop_duplicates(inplace=True)

print("Rows after duplicate removal:")
print(so_header.shape)
```

```python
print("Duplicate SO IDs:")
print(so_header['so_id'].duplicated().sum())

print("\nOrder Status Values:")
print(so_header['order_status'].unique())

print("\nSales Channel Values:")
print(so_header['sales_channel'].unique())
```

```python
so_header.to_csv('../data/processed/sales_orders_header.csv',
    index=False
)

print("Processed sales_orders_header.csv saved successfully!")
```
