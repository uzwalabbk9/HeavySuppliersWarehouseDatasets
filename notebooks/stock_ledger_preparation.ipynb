```python
import pandas as pd
stock_ledger = pd.read_csv('../data/raw/stock_ledger.csv')
print(stock_ledger.head())
```

```python
print("===INFO===")
print(stock_ledger.info())

print("===MISSING VALUES===")
print(stock_ledger.isnull().sum())

print("\n=== SHAPE ===")
print(stock_ledger.shape)

print("===SAMPLE DATA===")
print(stock_ledger.sample(5))

```

```python
text_cols = [
    'movement_id',
    'product_id',
    'branch_id',
    'movement_type',
    'reference_type',
    'reference_id'
]

for col in text_cols:
    stock_ledger[col] = stock_ledger[col].str.upper().str.strip()
```

```python
stock_ledger['movement_date'] = pd.to_datetime(
    stock_ledger['movement_date'],
    errors='coerce'
)

print(stock_ledger['movement_date'].head())
```

```python
numeric_cols = [
    'quantity',
    'running_balance'
]

print(stock_ledger[numeric_cols].describe())
```

```python
stock_ledger.drop_duplicates(inplace=True)

print("Rows after duplicate removal:")
print(stock_ledger.shape)
```

```python
print("Duplicate Movement IDs:")
print(stock_ledger['movement_id'].duplicated().sum())

print("\nMovement Types:")
print(stock_ledger['movement_type'].unique())

print("\nReference Types:")
print(stock_ledger['reference_type'].unique())
```

```python
stock_ledger.to_csv('../data/processed/stock_ledger.csv',
    index=False
)

print("Processed stock_ledger.csv saved successfully!")
```
