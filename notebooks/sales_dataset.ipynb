```python
import pandas as pd

so_header = pd.read_csv('../data/processed/sales_orders_header.csv')
so_lines = pd.read_csv('../data/processed/sales_orders_lines.csv')
customers = pd.read_csv('../data/processed/customers.csv')
products = pd.read_csv('../data/processed/products.csv')
```

```python
sales_curated = (
    so_lines
    .merge(so_header, on='so_id', how='left')
    .merge(customers, on='customer_id', how='left')
    .merge(products, on='product_id', how='left')
)

sales_curated.head()
```

```python
print("Shape:")
print(sales_curated.shape)

print("\nMissing Values:")
print(sales_curated.isnull().sum())
```

```python
sales_curated.to_csv(
    '../data/curated/sales_curated.csv',
    index=False
)

print("sales_curated.csv saved!")
```
