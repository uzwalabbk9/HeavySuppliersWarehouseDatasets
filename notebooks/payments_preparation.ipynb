```python
import pandas as pd
payments = pd.read_csv('../data/raw/payments.csv')
print(payments.head())
```

```python
print("===INFO===")
print(payments.info())

print("\n===Missing Values===")
print(payments.isnull().sum())

print("\n ===Sample Data===")
print(payments.head())
```

```python
text_cols = ['payment_id', 'invoice_id', 'payment_method']

for col in text_cols:
    payments[col] = payments[col].str.upper()
```

```python
payments['payment_date'] = pd.to_datetime(payments['payment_date'], errors='coerce')
payments[['payment_date']].head()
```

```python
payments['payment_year'] = payments['payment_date'].dt.year
payments['payment_month'] = payments['payment_date'].dt.month
payments['payment_quarter'] = payments['payment_date'].dt.quarter


payments['payment_weekday'] = payments['payment_date'].dt.day_name()

payments['high_value_flag'] = (payments['payment_amount'] > payments['payment_amount'].median()).astype(int)

payments['payment_method_category'] = payments['payment_method'].replace({
    'CREDIT CARD': 'CARD',
    'DEBIT CARD': 'CARD',
    'BANK TRANSFER': 'BANK',
    'CASH': 'CASH'
})
```

```python
payments.drop_duplicates(inplace=True)

```

```python
print("=== DUPLICATE PAYMENT IDs ===")
print(payments['payment_id'].duplicated().sum())

print("\n=== PAYMENT METHODS ===")
print(payments['payment_method'].unique())

print("\n=== DATE CHECK ===")
print(payments['payment_date'].head())

print("\n=== NUMERIC SUMMARY ===")
print(payments['payment_amount'].describe())
```

```python
payments.to_csv('../data/processed/payments.csv', index=False)
print("Processed payments.csv saved to data/processed/")
```
