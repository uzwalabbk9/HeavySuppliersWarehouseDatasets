```python
import pandas as pd

po_header = pd.read_csv('../data/processed/purchase_orders_header.csv')

po_lines = pd.read_csv('../data/processed/purchase_orders_lines.csv')

suppliers = pd.read_csv('../data/processed/suppliers.csv')

products = pd.read_csv('../data/processed/products.csv')
```

```python
procurement_curated = (
    po_lines
    .merge(po_header, on='po_id', how='left')
    .merge(suppliers, on='supplier_id', how='left')
    .merge(products, on='product_id', how='left')
)

procurement_curated.head()
```

```python
print("Shape:")
print(procurement_curated.shape)

print("\nMissing Values:")
print(procurement_curated.isnull().sum())
```

```python
procurement_curated.to_csv('../data/curated/procurement_curated.csv',
    index=False
)

print("procurement_curated.csv saved!")
```
