```python
import pandas as pd

inventory = pd.read_csv('../data/raw/inventory_master.csv')
inventory.head()
```

```python
print("=== INFO ===")
print(inventory.info())

print("\n=== MISSING VALUES ===")
print(inventory.isnull().sum())

print("\n=== SAMPLE DATA ===")
inventory.head()
```

```python
inventory['product_id'] = inventory['product_id'].str.upper()
inventory['branch_id'] = inventory['branch_id'].str.upper()
inventory['warehouse_bin'] = inventory['warehouse_bin'].str.upper()
```

```python
inventory['stock_availability_pct'] = (
    inventory['current_stock'] / inventory['max_stock']
)

```

```python
inventory['reorder_flag'] = (
    inventory['current_stock'] <= inventory['reorder_level']
).astype(int)
```

```python
inventory['safety_breach'] = (
    inventory['current_stock'] < inventory['safety_stock']
).astype(int)
```

```python
numeric_cols = [
    'opening_stock',
    'reorder_level',
    'safety_stock',
    'max_stock',
    'current_stock'
]




inventory.duplicated(subset=['product_id', 'branch_id']).sum()
inventory[numeric_cols].describe()
inventory['warehouse_bin'].unique()
```

```python
inventory.to_csv('../data/processed/inventory_master.csv', index=False)
```
