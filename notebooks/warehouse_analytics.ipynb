```python
import pandas as pd
import os

os.makedirs('../data/analytics', exist_ok=True)
```

```python
branches = pd.read_csv(
    '../data/processed/branches.csv'
)

inventory = pd.read_csv(
    '../data/processed/inventory_master.csv'
)

stock_ledger = pd.read_csv(
    '../data/processed/stock_ledger.csv'
)

po_header = pd.read_csv(
    '../data/processed/purchase_orders_header.csv'
)
```

```python
# Current Stock by Branch

# Warehouse Space Utilisation Analysis

branch_stock = (
    inventory.groupby('branch_id')
    ['current_stock']
    .sum()
    .reset_index()
)

# Merge with Branch Capacity
warehouse_utilisation = (
    branch_stock
    .merge(
        branches[
            ['branch_id',
             'warehouse_capacity']
        ],
        on='branch_id',
        how='left'
    )
)

# Utilisation %

warehouse_utilisation[
    'utilisation_pct'
] = (
    warehouse_utilisation['current_stock']
    /
    warehouse_utilisation['warehouse_capacity']
) * 100

warehouse_utilisation.head()


#Save the warehouse utilisation data to a CSV file
warehouse_utilisation.to_csv(
    '../data/analytics/warehouse_space_utilisation.csv',
    index=False
)
```

```python
# Warehouse Throughput Analysis

warehouse_throughput = (
    stock_ledger.groupby(
        'branch_id'
    )
    .size()
    .reset_index(name='total_movements')
)


warehouse_throughput.head()



# Save the warehouse throughput data to a CSV file


warehouse_throughput.to_csv(
    '../data/analytics/warehouse_throughput.csv',
    index=False
)
```

```python
# Warehouse Capacity Analysis

warehouse_capacity = (
    warehouse_utilisation.copy()
)

warehouse_capacity[
    'available_capacity'
] = (
    warehouse_capacity['warehouse_capacity']
    -
    warehouse_capacity['current_stock']
)

warehouse_capacity.head()

#save the warehouse capacity data to a CSV file

warehouse_capacity.to_csv(
    '../data/analytics/warehouse_capacity_analysis.csv',
    index=False
)

```

```python
# Warehouse Performance Scoring

warehouse_score = (
    warehouse_utilisation.copy()
)

warehouse_score[
    'performance_score'
] = (
    warehouse_score['utilisation_pct']
).round(2)

warehouse_score.head()

#save the warehouse performance score data to a CSV file
warehouse_score.to_csv(
    '../data/analytics/warehouse_performance_score.csv',
    index=False
)

```
