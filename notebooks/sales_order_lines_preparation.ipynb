```python
import pandas as pd
so_lines = pd.read_csv('../data/raw/sales_orders_lines.csv')
print(so_lines.head())
```

```python
print("===INFO===")
print(so_lines.info())

print("===MISSING VALUES===")
print(so_lines.isnull().sum())

print("\n=== SHAPE ===")
print(so_lines.shape)

print("===SAMPLE DATA===")
print(so_lines.sample(5))
```

```python
text_cols = ['so_id','product_id']

for col in text_cols:
    so_lines[col] = so_lines[col].str.upper()
```

```python
numeric_cols = [
    'line_number',
    'quantity',
    'unit_price',
    'gst_rate',
    'line_total',
    'gst_amount',
    'line_grand_total'
]

print(so_lines[numeric_cols].describe())
```

```python
so_lines.drop_duplicates(inplace=True)

print("Rows after duplicate removal:")
print(so_lines.shape)
```

```python
print("Duplicate Sales Order Lines:")
print(
    so_lines.duplicated(
        subset=['so_id', 'line_number']
    ).sum()
)

print("\nUnique Products:")
print(so_lines['product_id'].nunique())
```

```python
so_lines.to_csv('../data/processed/sales_orders_lines.csv',
    index=False
)

print(
    "Processed sales_orders_lines.csv saved successfully!"
)
```
