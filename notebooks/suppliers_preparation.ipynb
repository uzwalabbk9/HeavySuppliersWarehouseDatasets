```python
import pandas as pd
suppliers = pd.read_csv('../data/raw/suppliers.csv')
print(suppliers.head())
```

```python
print("===INFO====")
print(suppliers.info())

print("===MISSING VALUES===")
print(suppliers.isnull().sum())

print("\n=== SHAPE ===")
print(suppliers.shape)

print("===SAMPLE DATA===")
print(suppliers.sample(5))
```

```python
text_cols = [
    'supplier_id',
    'supplier_name',
    'supplier_type',
    'product_category',
    'city',
    'province',
    'region',
    'china_tax_id'
]

for col in text_cols:
    suppliers[col] = suppliers[col].str.upper().str.strip()
```

```python
numeric_cols = [
    'pincode',
    'lead_time_days',
    'reliability_score',
    'import_duty_rate'
]

print(suppliers[numeric_cols].describe())
```

```python
suppliers.drop_duplicates(inplace=True)

print("Rows after duplicate removal:")
print(suppliers.shape)
```

```python
print("Duplicate Supplier IDs:")
print(suppliers['supplier_id'].duplicated().sum())

print("\nSupplier Types:")
print(suppliers['supplier_type'].unique())

print("\nProduct Categories:")
print(suppliers['product_category'].unique())

print("\nRegions:")
print(suppliers['region'].unique())
```

```python
suppliers.to_csv('../data/processed/suppliers.csv',
    index=False
)

print("Processed suppliers.csv saved successfully!")
```
